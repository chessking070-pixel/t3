```python
f = open ('newfile.txt','r')
for i in f:
    words = i.split()
    initial = words[0]
    login = words[1]
    logout = words[2]
    login_hr = int(login[0:2]); login_min = int(login[-2:])
    logout_hr = int(logout[0:2]); logout_min = int(logout[-2:])
    diff = (logout_hr*60+logout_min) - (login_hr*60+login_min)
    if diff>=60:
        hr=diff//60
        minute=diff%60
        print(f"{initial} {login} {logout} ({hr}h {minute}m)")
```
