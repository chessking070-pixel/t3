### WAP to remove all the comments starting with a '#' from the given content in a txt file

```python
# txt file is 'data.txt'
f = open('data.txt','r')
g = open('final.txt','w+')
for i in f:
    if '#' in i:
        x = i.find('#')
        line = i[:x]
        g.write(line+'\n')
    else:
        g.write(i)
g.seek(0)
print(g.read())
f.close()
g.close()
```

### WAP that prompts that prompts user input of initial of the name and print then print the names that matches those initials from a given file namelist.txt

```python
f = open('namelist.txt','r')
user = input("Enter initial of name in 2 or 3 characters : ")
for i in f:
    words = i.split()
    if len(user) ==2:
        if user[0] == words[0][0] and user[-1] == words[-1][0]:
            print(i)
    elif len(user)==3:
        if len(words)==3:
            if user[0] == words[0][0] and user[1]==words[1][0] and user[2]==words[2][0]:
                print(i)
    else:
        print('invalid input')
f.close()
```
